## Disease Background: 

* Prevalence & incidence

Note: Many instances of Alzheimer’s research are closely linked with dementia, which is a general decline in mental ability. While they are not the same thing, Alzheimer’s causes 60-80% of dementia cases and both are known to be used interchangeably in some contexts.

The absolute incidence of Alzheimer’s is going down as time goes on. This is likely due to overall health and lifestyle changes like education and prevention/treatment of hypertension. However, the number of adults above the age of 65 is increasing every year such that overall cases of Alzheimer's are still increasing. It is estimated that around 7.2 million Americans age 65 or older are living with dementia, roughly 11% or 1 in 9. Autopsy studies and clinical trials have noted that 15-30% of patients exhibiting Alzheimer’s dementia lacked any Alzheimer’s related brain changes. Whether or not this should change the incidence is up to the individual context and use of this data in research studies or broader education.

NIH on Alzheimer's - https://pmc.ncbi.nlm.nih.gov/articles/PMC12040760/#alz70235-sec-0010


* Economic burden

There are both direct and indirect Alzheimer’s-related economic burdens on the people with the disease and their caretakers. Directly, there is the cost of medications, diagnostic tests, and direct care. Indirectly, there is a loss of income when the patient is no longer able to work, strain on family members that fall into the caretaker role, and decrease in income if those family members have to take on less hours to take care of the family member with Alzheimer’s. 

In 2015, total payments for Alzheimer’s care totaled ~$226 billion, and 15.7 million family members and other unpaid caregivers spent 17.9 billion hours providing care to people with Alzheimer’s. The unpaid care time may be lowered when the disease is diagnosed early, allowing family members to better plan for next steps as the disease progresses. 

Assessing the economic burden of Alzheimer’s disease patients first diagnosed by specialists - https://link.springer.com/article/10.1186/s12877-016-0303-5

Economic Burden of Alzheimer's Disease and Dementia Care - https://www.nature.com/nature-index/topics/l4/economic-burden-of-alzheimer-s-disease-and-dementia-care


* Risk factors (genetic, lifestyle)

Non-modifiable risk factors are currently the greatest indicators of Alzheimer’s in a given person. These being old age and a family history of Alzheimer’s dementia. Dementia affects 5% of people age 65-74, 13% of people 75-84, and 33% of people 85 and older. While it is the largest risk factor by far, old age alone does not cause dementia. Family history is specifically related to one gene, APOE which is responsible for a protein that transports cholesterol in the blood. The more copies of the e4 allele of the gene a person has, the more likely they are to develop Alzheimer’s. Risk of Alzheimer's roughly doubles for each allele of the gene. 

In terms of modifiable risk factors, more research is still needed to detangle causes of general dementia and causes of Alzheimer’s dementia, as being the leading cause of dementia, Alzheimer’s is often just assumed to be the cause. Regardless, modifiable risk factors for dementia are often long term, obesity, hypertension, and high cholesterol being a few examples. There are also cognitive factors like low education levels, head injuries, and vision loss. Smoking and excessive alcohol consumption are also shown to increase levels of dementia later in life. Many of these factors are closely tied to heart health as the brain is the largest use of blood oxygen in the entire body, an unhealthy heart is often the first step to cognitive deterioration.

* Societal determinants

Socioeconomic status (SES) may have relation to health, specifically when considering access to economic resources, income, education, employment, and social standing. Lower SES is related to less physical activity, higher diabetes risk, and other health risks and unhealthy habits that are risk factors for dementia. 

Social determinants of health (SDOH) are similar to socioeconomic status, but SDOH are non-medical factors, unlike some factors to do with SES. They are defined as the non-medical conditions in which people are born, grow, live, learn, work, and age. Key SDOH factors are education (associates lower education level achieved to poor brain health and more cognitive decline), access to healthcare (inconsistent healthcare visits and delayed diagnoses), physical environment (access to healthy food, clean air, and recreational spaces), economic stability (lack of nutrition in diets and safe living conditions), social connection (strong social networks can protect cognitive function), and social stress (accelerates biological aging). These can all contribute to the risk of Alzheimer’s disease.

* Symptoms

Common symptoms of dementia (caused by Alzheimer’s disease) are memory loss, communication problems, poor judgment, and changes to mood and behavior. These are all products of damaged communication between neurons and tend to start subtly and get worse over time. Patients can experience difficulty walking, speaking and swallowing in the late stages of the disease. Other cognitive symptoms include trouble understanding images, loss of spatial awareness, and confusion about the passage of time. Symptoms often occur years after initial brain decay and develop throughout several more years of life. Patients live on average 3-11 years after initial symptoms.

NIH on Alzheimer's - https://pmc.ncbi.nlm.nih.gov/articles/PMC12040760/#alz70235-sec-0010


* Diagnosis

There are three different stages in which Alzheimer’s disease may be diagnosed: early, middle, and late, also known as mild, moderate, and severe. 

Early-stage Alzheimer’s is typically noticed by close family and friends due to changes in behavior such as lapses in memory, inability to think of the right word, and remembering names. It can be diagnosed through a combination of cognitive assessments and medical evaluations. It starts with a doctor consulting with family members about changes in the patient’s behavior or personality, and blood and urine tests can be used to test for vitamin deficiencies and thyroid disorders that can mimic Alzheimer’s. Neuroimaging such as PET scans are the most important tools for detecting early-stage Alzheimer’s because they can be used to see brain atrophy, rule out abnormalities such as strokes or tumors, and detect amyloid plaques or tau accumulation which point to Alzheimer’s. Biomarker testing has emerged recently as a way to measure these beta-amyloid and tau protein levels, which again point to Alzheimer's. 

Middle-stage Alzheimer’s can be the longest stage, and this is when dementia symptoms are the most prominent. While the person can still participate in daily activities, they require assistance and may need tasks to be simplified. They also may need care beyond what family members can provide. Memory loss, facial recognition, and behavioral changes are all worsened in this stage. Standardized tests such as MMSE, MoCA, and AD8 are all cognitive and functional assessments that can determine if a patient meets dementia criteria. Physical and neurological examinations are conducted to assess motor skills, reflexes, and coordination. The rest of the tests are the same as early-stage diagnosis, and getting diagnosed at this stage does not fall on a single test but relies on the judgement of the clinician performing the tests.

Late-stage Alzheimer’s is accompanied with severe dementia symptoms and around-the-clock care. At this stage, the person may experience changes in physical abilities, communication, immune system, and loss of awareness of their surroundings. While they can still respond to interaction, they may not be able to initiate it. At this stage, DSM-5 and NIA-AA criteria diagnose Alzheimer’s if there is clear evidence of cognitive decline, loss of independence, consistency with Alzheimer’s disease pathology, and exclusion of other causes for dementia. At this stage, many people benefit from hospice to comfortably live out the rest of their life.

Alzheimer's Stages - Early, Middle, Late Dementia Symptoms - https://www.alz.org/alzheimers-dementia/stages


* Standard of care treatments (& reimbursement)

There is currently no cure for Alzheimer’s. However, there are several prescription drugs and therapies known to slow the progression of the disease. For early Alzheimer’s, cholinesterase inhibitors can be used to slow the breakdown of acetylcholine, a chemical believed to be important to memory and cognition. However as Alzheimer’s progresses, the body produces less and less acetylcholine and the treatment loses its effectiveness. For moderate Alzheimer’s, there are multiple FDA-approved drugs that target beta-amyloid in the brain. This can slow the buildup of plaques in the brain but not completely stop them. For severe Alzheimer’s, drugs can be administered that regulate other brain chemicals related to muscle function and antipsychosis to keep patients self-sufficient for a little bit longer. 

Many Alzheimer’s drugs are covered by insurance, Medicare, and Medicaid, depending on the dosage and method of delivery. Recently, a 2022 decision for Medicaid not to cover amyloid treatment for Alzheimer’s was overturned after massive public uproar. The turnover also included coverage for PET scans for the diagnosis and analysis of Alzheimer’s progression in patients. Typically, to be covered by any sort of payment plan, the drug needs to be FDA approved. Newer drugs and experimental therapies often require certain conditions to be met before coverage like cooperation in medical studies or being at a certain stage of progression.

CMS Treatment Coverage Policy: What It Means for People Living with Alzheimer’s - https://www.alz.org/alzheimers-dementia/treatments/cms-medicare-coverage

How Is Alzheimer's Disease Treated? - https://www.nia.nih.gov/health/alzheimers-treatment/how-alzheimers-disease-treated


* Disease progression & prognosis

__Stage 0: No evidence of clinical change, biomarkers are in normal range.__ 
At this stage while there are no noticeable symptoms, there are changes in the brain such as amyloid plaque buildup and tau changes.
There is no definitive long-term prognosis as some people with these brain changes do not develop dementia.

__Stage 1: Performance within expected range on cognitive tests, no cognitive decline or new symptoms.__
There are no significant cognitive changes but the brain changes continue to progress. The changes at this stage may also be detectable in biomarker tests such as PET scans or CSF tests. 
Prognosis is still inconclusive as developing dementia later in life is not guaranteed.

__Stage 2: Performance within expected range on cognitive tests but decline from previous levels, evidence of subtle long-term cognitive decline but still within normal range, may be accompanied by mood swings, fully independent.__
There are noticeable changes in memory and cognitive skill but the changes are still within the normal range, and not enough to take away the person’s independence.
Some people may progress to Alzheimer’s dementia while some return to normal cognitive levels.

__Stage 3: Performance is impaired outside of normal range, decline from baseline performance, independent performance of daily tasks but with increased difficulty.__
There are greater memory and cognitive changes, at this stage the person may begin to repeat questions, get lost, and not complete tasks as easily as before.
At this stage the disease will progressively decline and continue to interfere with everyday life.

__Stage 4: Progressive cognitive decline and mild functional impairment.__
Memory loss gets worse and the person begins to have trouble with facial recognition, remembering new information, and completing tasks.
Loss of some independence, more dependence on caregivers for basic everyday tasks.

__Stage 5: Progressive cognitive and moderate functional impairment.__
Memory loss continues to worsen, the person may be confused, and have trouble completing tasks. They may also exhibit a change in behavior or personality.
They have increased dependence on caregivers at this stage and need help with basic tasks.

__Stage 6: Progressive cognitive and functional impairment, complete dependence.__
The person has severe cognitive and functional impairment and can no longer communicate. They rely entirely on others for daily activities. 
They require around-the-clock care and no longer maintain all of their mobility.

* Continuum of care providers

Care providers are often friends and family close to the patient. 83% of which are unpaid. Caregiving tasks often include helping with instrumental activities of daily life like shopping, cooking, managing dates and schedules, transportation, managing medications, bathing, dressing, and using the bathroom. Caregivers also provide necessary emotional and social support for Alzheimer’s patients, communicating with family members and other caregivers. 

Nursing homes and long term living arrangements are also very common, with over 46% of nursing home residents having Alzheimer’s or dementia. At age 80, approximately 75% of people with Alzheimer’s dementia live in nursing homes. These homes provide daily care of medications, personal help with aforementioned daily activities of life, and contact with friends and family. There are also Alzheimer’s special care units which are dedicated floors or wings in a nursing home that have dedicated services for people with Alzheimer’s or other dementias. Roughly 13% of nursing homes have dedicated Alzheimer’s care units.

NIH on Alzheimer's - https://pmc.ncbi.nlm.nih.gov/articles/PMC12040760/#alz70235-sec-0010


* Biological mechanisms (anatomy, organ physiology, cell & molecular physiology)

Alzheimer’s disease is thought to be caused by amyloid-beta (Aβ) plaques and tau protein tangles in the brain that lead to neuron death and synapse loss. The main brain structures affected are the hippocampus and cerebral cortex. The hippocampus forms and retrieves memories, and the cerebral cortex is responsible for thinking, decision making, language, etc. Neurons typically transmit information at synapses via electrical and chemical signals, but Alzheimer’s causes synaptic loss which slows the information flow, leading to cognitive decline. The Aβ plaques and tau tangles are what disrupt the synapses and impair the information transport.

What are the biological mechanisms of Alzheimer's disease? - https://parisbraininstitute.org/disease-files/alzheimers-disease/what-are-biological-mechanisms-alzheimers-disease

Mechanisms of Alzheimer’s Disease Pathogenesis and Prevention: The Brain, Neural Pathology, N-methyl-D-aspartate Receptors, Tau Protein and Other Risk Factors - https://pmc.ncbi.nlm.nih.gov/articles/PMC5290713/ 


* Clinical Trials/next-gen therapies

With the total cases of Alzheimer’s increasing as the average person lives longer and longer, the race to find a cure to Alzheimer’s is getting more important every day. Clinical trials are being held all over the country and are focusing on a wide variety of treatments and methods of understanding. Blood pressure affects memory, nasal insulin as a tool against early symptoms, molecular imaging to study how amyloid and tau build up over time, and electroconvulsive therapy to treat anger and depression, just to name a few. Many trials also focus on specific drugs like Lorazepam, CAD106, or MK-8931 to investigate their viability in slowing the disease. These trials are in varying stages of success and development and many of them will fail, as most research does. But failure is a tool for the people developing these next-gen therapies and their success is built on the failure of thousands of people before them until one team finally cracks the case. More therapies are coming out every year and this is the part of Alzheimer’s research that is always changing the fastest.

Mayo Clinic Clinical Trials - https://www.mayo.edu/research/clinical-trials/diseases-conditions/alzheimer%27s-disease
